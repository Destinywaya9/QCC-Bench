```python
# QCC Echo Imports
from qiskit import QuantumCircuit, transpile
from qiskit_aer import Aer # Import Aer from qiskit_aer
from qiskit_aer.noise import NoiseModel, depolarizing_error
import numpy as np
import matplotlib.pyplot as plt
import math
```

```python
# Construct QCC Echo Kernel (Abstracted Sample Form)
qc = QuantumCircuit(2, 2)
qc.h(0)
qc.cx(0, 1)
qc.barrier()

# Stabilizing Echo Operation
qc.cx(0, 1)
qc.h(0)

# Measurement
qc.measure([0, 1], [0, 1])
```

```python
# Define depolarizing noise model with 0.0 error (ideal coherence)
noise_model = NoiseModel()
error = depolarizing_error(0.0, 1)
noise_model.add_all_qubit_quantum_error(error, ['u1', 'u2', 'u3', 'cx'])
```

```python
# Execute the circuit on the QASM simulator
# backend = Aer.get_backend('qasm_simulator') # Deprecated way to get backend
from qiskit_aer import AerSimulator # Import AerSimulator
simulator = AerSimulator(noise_model=noise_model) # Instantiate AerSimulator with noise model

# job = execute(qc, backend=backend, shots=8192, noise_model=noise_model) # Deprecated execute function
transpiled_qc = transpile(qc, simulator) # Transpile the circuit
job = simulator.run(transpiled_qc, shots=8192) # Use simulator.run()
result = job.result()

# Get the result counts
counts = result.get_counts(qc)
print("Counts:", counts)
```

```python
# Plot the histogram of measurement outcomes
from qiskit.visualization import plot_histogram
plot_histogram(counts)
```

```python
import math

# Entropy Calculation
def shannon_entropy(probabilities):
    entropy = 0.0
    for p in probabilities:
        if p > 0:
            entropy -= p * math.log2(p)
    return entropy

# Extract probability distribution
total_shots = sum(counts.values())
probabilities = [v / total_shots for v in counts.values()]

# Entropy in bits
entropy = shannon_entropy(probabilities)

# Coherence Fidelity Calculation
max_prob = max(probabilities)
coherence_fidelity = max_prob * 100

print(f"Entropy Leakage: {entropy:.6f} bits")
print(f"Coherence Fidelity: {coherence_fidelity:.8f}%")
```

```python
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator
from qiskit_aer.noise import NoiseModel, depolarizing_error
from qiskit.visualization import plot_histogram
import matplotlib.pyplot as plt
import numpy as np

# --- Define EchoP2 Quantum Circuit (QCC-Kernel Locked) ---
qc = QuantumCircuit(3, 3)

# --- QCC-style Resonance Prep ---
qc.h(0)
qc.cx(0, 1)
qc.cx(1, 2)

# --- QCC Echo Padding (mirrored gates) ---
qc.cx(1, 2)
qc.cx(0, 1)
qc.h(0)

# --- QCC Rhythm Buffer ---
qc.barrier()
qc.delay(200, 0)
qc.delay(200, 1)
qc.delay(200, 2)

# --- Measurement ---
qc.measure([0, 1, 2], [0, 1, 2])

# --- Noise Model Injection (0.1 depolarizing noise) ---
noise_model = NoiseModel()
# Create 1-qubit and 2-qubit depolarizing errors
error_1q = depolarizing_error(0.1, 1)
error_2q = depolarizing_error(0.1, 2) # Create a 2-qubit error

# Add errors to the noise model for specific instructions
noise_model.add_all_qubit_quantum_error(error_1q, ['h'])
noise_model.add_all_qubit_quantum_error(error_2q, ['cx'])


# --- Simulator Execution ---
simulator = AerSimulator(noise_model=noise_model)
transpiled_qc = transpile(qc, simulator)
job = simulator.run(transpiled_qc, shots=8192)
result = job.result()

# --- Output Histogram ---
counts = result.get_counts()
plot_histogram(counts, title="QCC Echo P2 with 0.1 Depolarizing Noise")
plt.show()

# --- Entropy & Coherence Calculation ---
probs = np.array(list(counts.values())) / 8192
entropy = -np.sum([p * np.log2(p) for p in probs if p > 0])
max_prob = max(probs)
coherence_fidelity = max_prob * 100

print(f"Entropy Leakage: {entropy:.6f} bits")
print(f"Coherence Fidelity: {coherence_fidelity:.8f}%")
```

```python

# Install and import dependencies
!pip install qiskit qiskit-aer matplotlib --quiet

import numpy as np
import matplotlib.pyplot as plt
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator
from qiskit_aer.noise import NoiseModel, depolarizing_error, thermal_relaxation_error
from qiskit.visualization import plot_histogram

# --- Define QCC Echo Kernel ---
qc = QuantumCircuit(3, 3)
qc.h(0)
qc.cx(0, 1)
qc.cx(1, 2)
qc.cx(1, 2)
qc.cx(0, 1)
qc.h(0)
qc.barrier()
qc.delay(200, 0)
qc.delay(200, 1)
qc.delay(200, 2)
qc.measure([0, 1, 2], [0, 1, 2])

# --- Noise Model Setup ---
T1 = 50e3  # nanoseconds
T2 = 30e3  # nanoseconds
gate_time = 100  # ns

noise_model = NoiseModel()
for qubit in [0, 1, 2]:
    thermal_error = thermal_relaxation_error(T1, T2, gate_time)
    noise_model.add_quantum_error(thermal_error, ['u1', 'u2', 'u3'], [qubit])

depol_error = depolarizing_error(0.02, 2)
noise_model.add_quantum_error(depol_error, 'cx', [0, 1])
noise_model.add_quantum_error(depol_error, 'cx', [1, 2])

# --- Simulators ---
sim_clean = AerSimulator()  # No noise
sim_noisy = AerSimulator(noise_model=noise_model)

# --- Transpile Circuits ---
qc_clean = transpile(qc, sim_clean)
qc_noisy = transpile(qc, sim_noisy)

# --- Run Simulations ---
result_clean = sim_clean.run(qc_clean, shots=8192).result()
result_noisy = sim_noisy.run(qc_noisy, shots=8192).result()

counts_clean = result_clean.get_counts()
counts_noisy = result_noisy.get_counts()

# --- Plot Results ---
plot_histogram([counts_clean, counts_noisy], legend=["Clean", "Noise"], title="QCC Echo Dual Run")
plt.show()

# --- Entropy & Coherence for Noisy Run ---
probs = np.array(list(counts_noisy.values())) / sum(counts_noisy.values())
entropy = -np.sum([p * np.log2(p) for p in probs if p > 0])
coherence_fidelity = max(probs) * 100

print(f"Entropy Leakage (Noisy): {entropy:.6f} bits")
print(f"Coherence Fidelity (Noisy): {coherence_fidelity:.8f}%")
```
