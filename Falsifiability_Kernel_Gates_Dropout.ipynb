```python
# Install Qiskit and qiskit-aer
!pip install qiskit qiskit-aer --quiet

# Import core tools
from qiskit import QuantumCircuit
from qiskit_aer import Aer  # Import Aer from qiskit_aer
from qiskit.visualization import plot_histogram
import matplotlib.pyplot as plt

# QCC-Compatible: Simple Bell State Circuit (proof-of-coherence gate)
qc = QuantumCircuit(2, 2)
qc.h(0)           # Hadamard on qubit 0
qc.cx(0, 1)       # CNOT with qubit 0 as control, 1 as target
qc.measure([0,1], [0,1])

# Run simulation
simulator = Aer.get_backend('qasm_simulator')
job = simulator.run(qc, shots=1024) # Use the run method
result = job.result()
counts = result.get_counts(qc)

# Output result
print("Counts:", counts)
plot_histogram(counts)
plt.show()
```
